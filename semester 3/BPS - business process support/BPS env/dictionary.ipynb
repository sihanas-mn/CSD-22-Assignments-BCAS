```python
new_dic = {
    'StudentID': [47, 14, 78, 45, 15],
    'Name': ["Alice", "Bob", "Charli", "David", "Eve"],
    'PCP Score': [95, 85, 75, 65, 55],
    'BPS Score': [85, 75, 65, 55, 45],
    'DDD Score': [180, 160, 140, 120, 100]
}

```

```python
new_df = pd.DataFrame(new_dic)
new_df
```

```python
new_df.sort_values(by='StudentID', inplace=True)
new_df
```

```python
third_df = new_df.sort_values(by=["StudentID", "Name"], ascending=[False, False])
```

```python
third_df
```

```python
third_df['Total Score'] = third_df['PCP Score'] + third_df['BPS Score'] + third_df['DDD Score']
```
