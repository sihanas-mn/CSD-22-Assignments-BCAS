```python
my_list = [11, 52, 33, 44, 55, 16, 32]
print(my_list)
```

```python
even_numbers = [num for num in my_list if num % 2 == 0]
print(even_numbers)
```

```python
print(my_list[1 : 4 : 2])
```

```python
even_numbers = []

for num in my_list:
    if num % 2 == 0:
        even_numbers.append(num)
    
print(even_numbers)
```

```python
even_numbers = []
odd_numbers = []
for num in my_list:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)
print("Here are the odd numbers from you list: ", odd_numbers)
print("Here are the even numbers from you list: ", even_numbers)
```
