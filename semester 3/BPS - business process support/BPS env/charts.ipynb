```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
city = np.array(['Beijing', 'Shanghai', 'Guangzhou', 'Shenzhen', 'Hangzhou'])
population = np.array([35, 51, 86, 16, 21])
```

```python
mycolor = sns.color_palette("icefire")
plt.figure(figsize=(5, 4))
plt.bar(city, population, color=mycolor)
plt.xlabel('City')
plt.ylabel('Population in million')
plt.title('Population of Major Cities in China')
plt.show()
```

```python
# Read the CSV file
sales_df = pd.read_csv('sales_data.csv')

# Group by city and sum the totals
grouped_sales = sales_df.groupby('City')['Total'].sum()

mycolor2 = sns.color_palette("rocket_r")
# Create horizontal bar chart
plt.figure(figsize=(5, 4))
plt.bar(grouped_sales.index, grouped_sales.values, color=mycolor2[:len(grouped_sales)])
plt.xlabel('City')
plt.ylabel('Total')
plt.title('Sales by City')
plt.show()
```

```python
# Sort the grouped sales in descending order
sorted_sales = grouped_sales.sort_values(ascending=True)

mycolor3 = sns.color_palette("viridis")
plt.figure(figsize=(5, 4))
plt.bar(sorted_sales.index, sorted_sales.values, color=mycolor3[:len(sorted_sales)])
plt.xlabel('City')
plt.ylabel('Total')
plt.title('Sales by City (Sorted)')
plt.xticks(rotation=90)
plt.show()
```

```python
# Read the CSV file
quantity_df = pd.read_csv('sales_data.csv')

# Group by city and sum the totals
grouped_quantity = quantity_df.groupby('Product line')['Quantity'].sum()

mycolor2 = sns.color_palette("rocket_r")
# Create horizontal bar chart
plt.figure(figsize=(6, 5))
plt.bar(grouped_quantity.index, grouped_quantity.values, color=mycolor2[:len(grouped_quantity)])
plt.xlabel('Product line')
plt.ylabel('Quantity')
plt.title('product by quantity')
plt.xticks(rotation=90)
plt.show()
```

```python
# Sort the grouped quantity in descending order
sorted_quantity = grouped_quantity.sort_values(ascending=True)

plt.figure(figsize=(6, 5))
plt.bar(sorted_quantity.index, sorted_quantity.values, color=mycolor2[:len(sorted_quantity)])
plt.xlabel('Product line')
plt.ylabel('Quantity')
plt.title('Product by Quantity (Sorted)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
gender_counts = sales_df['Gender'].value_counts()

plt.figure(figsize=(5, 5))
plt.pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%', 
    colors=mycolor[:len(gender_counts)], startangle=90)
plt.title('Customer Gender Distribution')
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(5, 6))
plt.pie(gender_counts.values, autopct='%1.1f%%', 
    colors=mycolor[:len(gender_counts)], startangle=90)
plt.title('Customer Gender Distribution')
plt.legend(gender_counts.index, title="Gender", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))
plt.axis('equal')
plt.show()
```
