```python
"""
YOLOv8 Training and Evaluation Script
This script handles the complete workflow for training and evaluating YOLOv8-Small
on a road traffic sign dataset, with automatic checkpointing of the best model
and robust evaluation metrics.
"""

import os
import sys
import zipfile
import shutil
import time
import csv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
from pathlib import Path
from google.colab import drive
import torch
import yaml

# Check if running in Google Colab and using GPU
def check_environment():
    """Check if running in Colab with GPU."""
    is_colab = 'google.colab' in sys.modules
    has_gpu = torch.cuda.is_available()
    gpu_type = torch.cuda.get_device_name(0) if has_gpu else "None"

    print(f"Running in Google Colab: {is_colab}")
    print(f"GPU Available: {has_gpu}")
    print(f"GPU Type: {gpu_type}")

    if not is_colab:
        print("Warning: This script is designed to run in Google Colab.")

    if not has_gpu:
        print("Warning: No GPU detected. Training will be very slow.")
    elif "T4" not in gpu_type:
        print(f"Note: Using {gpu_type} instead of Tesla T4.")

    return is_colab, has_gpu

# Mount Google Drive
def mount_drive():
    """Mount Google Drive and return the mounted path."""
    print("Mounting Google Drive...")
    drive.mount('/content/drive')
    return '/content/drive/MyDrive/'

# Install dependencies
def install_dependencies():
    """Install required packages."""
    print("Installing dependencies...")
    # Install Ultralytics for YOLOv8
    !pip install -q ultralytics
    # Additional dependencies if needed
    !pip install -q opencv-python-headless

    # Import after installation to avoid errors
    from ultralytics import YOLO
    print("Dependencies installed successfully.")
    return YOLO

# Extract dataset from zip
def extract_dataset(zip_path, extract_to):
    """Extract the dataset from zip file."""
    print(f"Extracting dataset from {zip_path}...")
    os.makedirs(extract_to, exist_ok=True)

    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)

    # Validate extraction
    yaml_path = os.path.join(extract_to, 'data.yaml')
    if not os.path.exists(yaml_path):
        print("Warning: data.yaml not found in extracted dataset.")
        # Try to find it in subfolders
        for root, dirs, files in os.walk(extract_to):
            if 'data.yaml' in files:
                yaml_path = os.path.join(root, 'data.yaml')
                print(f"Found data.yaml at: {yaml_path}")
                break

    return yaml_path

# Create result directories
def create_output_dirs(base_path):
    """Create directory structure for outputs."""
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    output_dir = os.path.join(base_path, f"/content/drive/MyDrive/YOLOv8_Results_{timestamp}")
    os.makedirs(output_dir, exist_ok=True)

    # Create subdirectories
    plots_dir = os.path.join(output_dir, "plots")
    os.makedirs(plots_dir, exist_ok=True)

    return output_dir, plots_dir

# Custom training loop with early stopping
def train_model(data_yaml_path, output_dir, epochs=20, patience=10):
    """
    Train YOLOv8 model with early stopping and best model saving.

    Args:
        data_yaml_path: Path to data.yaml file
        output_dir: Directory to save results
        epochs: Number of training epochs
        patience: Early stopping patience (epochs without improvement)

    Returns:
        Path to best model and training metrics
    """
    from ultralytics import YOLO

    # Load a pre-trained YOLOv8s model
    model = YOLO('yolov8s.pt')

    # Initialize tracking variables
    best_map50 = 0
    best_epoch = 0
    no_improve_count = 0
    best_model_path = os.path.join(output_dir, 'best.pt')

    # Dictionaries to store metrics
    metrics_history = {
        'epoch': [],
        'train_loss': [],
        'val_loss': [],
        'mAP50': [],
        'mAP50-95': [],
        'precision': [],
        'recall': [],
        'f1': []
    }

    # Check data.yaml and fix paths if needed
    validate_and_fix_yaml(data_yaml_path)

    print(f"Starting training for {epochs} epochs...")
    print(f"Early stopping patience: {patience} epochs")

    # Train the model for all epochs at once (more efficient)
    # We'll validate after each epoch using model.val()
    model.train(
        data=data_yaml_path,
        epochs=epochs,
        imgsz=640,
        batch=16,
        device=0,
        exist_ok=True,
        project=output_dir,
        name='training',
        plots=True,  # Save training plots
        save=True    # Save checkpoints
    )

    # After training, locate the best model weight file
    training_dir = os.path.join(output_dir, 'training')
    weights_dir = os.path.join(training_dir, 'weights')

    if os.path.exists(os.path.join(weights_dir, 'best.pt')):
        # Copy best model to output directory
        shutil.copy(os.path.join(weights_dir, 'best.pt'), best_model_path)
        print(f"Best model copied to: {best_model_path}")
    else:
        # Use the last model if best is not available
        shutil.copy(os.path.join(weights_dir, 'last.pt'), best_model_path)
        print(f"Last model copied to: {best_model_path} (best not found)")

    # Load training results from CSV
    try:
        results_csv = os.path.join(training_dir, 'results.csv')
        if os.path.exists(results_csv):
            results_df = pd.read_csv(results_csv)

            # Convert results to our metrics format
            metrics_history['epoch'] = list(range(1, len(results_df) + 1))
            metrics_history['train_loss'] = results_df['train/box_loss'].tolist() if 'train/box_loss' in results_df.columns else [np.nan] * len(results_df)
            metrics_history['val_loss'] = results_df['val/box_loss'].tolist() if 'val/box_loss' in results_df.columns else [np.nan] * len(results_df)
            metrics_history['mAP50'] = results_df['metrics/mAP50(B)'].tolist() if 'metrics/mAP50(B)' in results_df.columns else [np.nan] * len(results_df)
            metrics_history['mAP50-95'] = results_df['metrics/mAP50-95(B)'].tolist() if 'metrics/mAP50-95(B)' in results_df.columns else [np.nan] * len(results_df)

            # Find precision and recall columns (may vary based on version)
            for col in results_df.columns:
                if 'precision' in col.lower():
                    metrics_history['precision'] = results_df[col].tolist()
                    break
            else:
                metrics_history['precision'] = [np.nan] * len(results_df)

            for col in results_df.columns:
                if 'recall' in col.lower():
                    metrics_history['recall'] = results_df[col].tolist()
                    break
            else:
                metrics_history['recall'] = [np.nan] * len(results_df)

            # Calculate F1 from precision and recall
            metrics_history['f1'] = [
                2 * (p * r) / (p + r) if not (np.isnan(p) or np.isnan(r) or p + r == 0) else np.nan
                for p, r in zip(metrics_history['precision'], metrics_history['recall'])
            ]

            # Find the best epoch
            if not all(np.isnan(metrics_history['mAP50'])):
                best_index = np.nanargmax(metrics_history['mAP50'])
                best_map50 = metrics_history['mAP50'][best_index]
                best_epoch = best_index + 1
                print(f"Best model found at epoch {best_epoch} with mAP50: {best_map50:.4f}")

        else:
            print("Warning: results.csv not found. Using empty metrics.")
            # Generate empty metrics for plotting
            metrics_history = {k: [np.nan] * epochs for k in metrics_history.keys()}
            metrics_history['epoch'] = list(range(1, epochs + 1))

    except Exception as e:
        print(f"Error processing training results: {e}")
        # Create dummy metrics for plotting
        metrics_history = {k: [np.nan] * epochs for k in metrics_history.keys()}
        metrics_history['epoch'] = list(range(1, epochs + 1))

    # Create a DataFrame from the metrics
    metrics_df = pd.DataFrame(metrics_history)
    metrics_csv_path = os.path.join(output_dir, 'training_metrics.csv')
    metrics_df.to_csv(metrics_csv_path, index=False)

    print(f"\nTraining completed. Metrics saved to {metrics_csv_path}")

    return best_model_path, metrics_df, best_epoch

# Validate and potentially fix data.yaml
def validate_and_fix_yaml(yaml_path):
    """Validate and fix paths in data.yaml if needed."""
    with open(yaml_path, 'r') as f:
        data = yaml.safe_load(f)

    # Get the directory where data.yaml is located
    base_dir = os.path.dirname(yaml_path)

    # Check if paths are relative and exist
    modified = False
    for key in ['train', 'val', 'test']:
        if key in data:
            # Create absolute path if relative
            if not os.path.isabs(data[key]):
                abs_path = os.path.join(base_dir, data[key])
                if os.path.exists(abs_path):
                    data[key] = abs_path
                    modified = True

    # Write back if modified
    if modified:
        print("Updating data.yaml with absolute paths...")
        with open(yaml_path, 'w') as f:
            yaml.dump(data, f)

    return data

# Evaluate best model on test set
def evaluate_model(model_path, data_yaml_path, output_dir):
    """Evaluate the best model on the test set."""
    from ultralytics import YOLO

    print(f"\nEvaluating best model from {model_path} on test set...")

    # Load the best model
    model = YOLO(model_path)

    # Run evaluation on test set
    results = model.val(data=data_yaml_path, split='test')

    # Save metrics
    metrics = {
        "mAP50": results.box.map50,
        "mAP50-95": results.box.map,
        "Precision": results.box.mp,  # mean precision
        "Recall": results.box.mr,     # mean recall
        "F1-Score": results.box.f1.mean() if isinstance(results.box.f1, (list, np.ndarray)) else results.box.f1
    }

    # Save metrics to CSV
    metrics_path = os.path.join(output_dir, 'final_metrics.csv')
    with open(metrics_path, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['Metric', 'Value'])
        for key, value in metrics.items():
            writer.writerow([key, f"{value:.4f}"])

    print("Final evaluation metrics:")
    for key, value in metrics.items():
        print(f"{key}: {value:.4f}")

    return metrics

# Generate performance plots
def generate_plots(metrics_df, plots_dir):
    """Generate and save performance plots."""
    # Plot mAP50 over epochs
    plt.figure(figsize=(10, 6))
    plt.plot(metrics_df['epoch'], metrics_df['mAP50'], 'b-', label='mAP50')
    plt.plot(metrics_df['epoch'], metrics_df['mAP50-95'], 'g--', label='mAP50-95')
    plt.xlabel('Epoch')
    plt.ylabel('mAP')
    plt.title('Mean Average Precision (mAP) per Epoch')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(plots_dir, 'mAP_plot.png'))

    # Plot F1 score over epochs
    plt.figure(figsize=(10, 6))
    plt.plot(metrics_df['epoch'], metrics_df['f1'], 'r-', label='F1-Score')
    plt.plot(metrics_df['epoch'], metrics_df['precision'], 'g--', label='Precision')
    plt.plot(metrics_df['epoch'], metrics_df['recall'], 'b--', label='Recall')
    plt.xlabel('Epoch')
    plt.ylabel('Score')
    plt.title('F1 Score, Precision and Recall per Epoch')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(plots_dir, 'F1_plot.png'))

    # Plot loss over epochs (skip if data is not available)
    if not all(np.isnan(metrics_df['train_loss'])):
        plt.figure(figsize=(10, 6))
        plt.plot(metrics_df['epoch'], metrics_df['train_loss'], 'r-', label='Train Loss')
        if not all(np.isnan(metrics_df['val_loss'])):
            plt.plot(metrics_df['epoch'], metrics_df['val_loss'], 'b-', label='Validation Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.title('Loss per Epoch')
        plt.legend()
        plt.grid(True)
        plt.savefig(os.path.join(plots_dir, 'loss_plot.png'))
    else:
        print("Skipping loss plot as loss data is not available")

    print(f"Performance plots saved to {plots_dir}")

# Copy results to Drive
def save_to_drive(source_dir, drive_path):
    """Copy results to Google Drive."""
    drive_output = os.path.join(drive_path, os.path.basename(source_dir))

    # Create the directory if it doesn't exist
    os.makedirs(drive_output, exist_ok=True)

    # Copy all files from source_dir to drive_output
    for item in os.listdir(source_dir):
        s = os.path.join(source_dir, item)
        d = os.path.join(drive_output, item)
        if os.path.isdir(s):
            shutil.copytree(s, d, dirs_exist_ok=True)
        else:
            shutil.copy2(s, d)

    print(f"Results saved to Google Drive: {drive_output}")
    return drive_output

# Main function
def main():
    """Main execution function."""
    # Configuration parameters - adjust as needed
    EPOCHS = 100
    PATIENCE = 10  # Early stopping patience
    ZIP_FILENAME = "Road Traffic Sign In Sri Lanka.v4i.yolov8.zip"
    DRIVE_RESULTS_FOLDER = "YOLOv8_Results"

    # Check environment
    is_colab, has_gpu = check_environment()

    # Mount Google Drive
    drive_path = mount_drive() if is_colab else "./"

    # Install dependencies
    YOLO = install_dependencies()

    # Setup paths
    zip_path = os.path.join(drive_path, ZIP_FILENAME)
    extract_path = '/content/dataset'

    # Create output directories
    output_dir, plots_dir = create_output_dirs('/content')

    # Extract dataset
    data_yaml_path = extract_dataset(zip_path, extract_path)
    print(f"Dataset extracted to {extract_path}")
    print(f"Using data.yaml from: {data_yaml_path}")

    # Train model
    best_model_path, metrics_df, best_epoch = train_model(
        data_yaml_path=data_yaml_path,
        output_dir=output_dir,
        epochs=EPOCHS,
        patience=PATIENCE
    )

    # Evaluate best model
    final_metrics = evaluate_model(best_model_path, data_yaml_path, output_dir)

    # Generate plots
    generate_plots(metrics_df, plots_dir)

    # Save results to Drive
    if is_colab:
        drive_results_path = os.path.join(drive_path, DRIVE_RESULTS_FOLDER)
        save_to_drive(output_dir, drive_results_path)

    print("\nComplete YOLOv8 training and evaluation workflow finished successfully!")
    print(f"Best model found at epoch {best_epoch} with mAP50: {final_metrics['mAP50']:.4f}")
    print(f"All results saved to: {output_dir}")

if __name__ == "__main__":
    main()
```
