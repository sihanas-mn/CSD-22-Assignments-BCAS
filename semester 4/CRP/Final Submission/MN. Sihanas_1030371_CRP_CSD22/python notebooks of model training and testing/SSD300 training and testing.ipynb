```python
import os
import sys
import zipfile
import time
import random
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision
from torchvision import transforms
from torchvision.models.detection.ssd import SSD300_VGG16_Weights
import xml.etree.ElementTree as ET
from tqdm.notebook import tqdm
import pandas as pd
from google.colab import drive
from PIL import Image, ImageDraw
import cv2

# Check if GPU is available
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# If using Colab, check for T4 GPU
if device.type == 'cuda':
    gpu_info = !nvidia-smi
    print("GPU Info:")
    for line in gpu_info:
        print(line)
    if "T4" not in '\n'.join(gpu_info):
        print("Warning: T4 GPU not detected. Performance may be affected.")

# Mount Google Drive
print("Mounting Google Drive...")
drive.mount('/content/drive')

# Configuration parameters - adjust as needed
CONFIG = {
    'dataset_zip_path': '/content/drive/MyDrive/Road Traffic Sign In Sri Lanka.v4i.voc.zip',
    'extraction_dir': '/content/traffic_signs_dataset',
    'results_dir': '/content/drive/MyDrive/SSD300_Results',
    'epochs': 30,
    'batch_size': 16,
    'learning_rate': 0.001,
    'early_stopping_patience': 5,
    'img_size': 300,  # SSD300 input size
    'num_workers': 4
}

# Create results directory if it doesn't exist
os.makedirs(CONFIG['results_dir'], exist_ok=True)

# Class names for Sri Lankan road traffic signs
CLASS_NAMES = ['Bus-stop', 'Compulsory-Roundabout', 'Cross-Roads-Ahead',
               'Double-Bend-to-Left-Ahead', 'Double-Bend-to-Right-Ahead',
               'Falling-Rocks-Ahead', 'Left-Bend-Ahead',
               'Level-crossing-with-barriers-ahead',
               'Level-crossing-without-barriers-ahead',
               'Narrow-Bridge-or-Culvert-Ahead', 'No-entry', 'No-horns',
               'No-left-turn', 'No-parking', 'No-parking-and-standing',
               'No-parking-on-even-numbered-days',
               'No-parking-on-odd-numbered-days', 'No-right-turn',
               'No-u-turn', 'Pedestrian-Crossing', 'Pedestrian-crossing-Ahead',
               'Proceed-straight', 'Right-Bend-Ahead', 'Road-Bump-Ahead',
               'Road-Closed-for-All-Vehicles', 'Roundabout-Ahead',
               'Speed-Limit-40-Kmph', 'Speed-Limit-50-Kmph',
               'Speed-Limit-60-Kmph', 'Stop', 'T-Junction-Ahead',
               'Traffic-from-left-merges-ahead',
               'Traffic-from-right-merges-ahead', 'Turn-left',
               'Turn-left-ahead', 'Turn-right', 'Turn-right-ahead']

# Install necessary packages if not already installed
!pip install -q pycocotools

# Extract dataset
print(f"Extracting dataset from {CONFIG['dataset_zip_path']}...")
with zipfile.ZipFile(CONFIG['dataset_zip_path'], 'r') as zip_ref:
    zip_ref.extractall(CONFIG['extraction_dir'])

print(f"Dataset extracted to {CONFIG['extraction_dir']}")

# Verify extraction and dataset structure
train_dir = os.path.join(CONFIG['extraction_dir'], 'train')
valid_dir = os.path.join(CONFIG['extraction_dir'], 'valid')
test_dir = os.path.join(CONFIG['extraction_dir'], 'test')

print(f"Train directory exists: {os.path.exists(train_dir)}")
print(f"Validation directory exists: {os.path.exists(valid_dir)}")
print(f"Test directory exists: {os.path.exists(test_dir)}")

# Count files
train_jpgs = len([f for f in os.listdir(train_dir) if f.endswith('.jpg')])
train_xmls = len([f for f in os.listdir(train_dir) if f.endswith('.xml')])
print(f"Train images: {train_jpgs}, Train annotations: {train_xmls}")

# Custom transformations for object detection
class Compose:
    def __init__(self, transforms):
        self.transforms = transforms

    def __call__(self, image, target):
        for t in self.transforms:
            image, target = t(image, target)
        return image, target

class ToTensor:
    def __call__(self, image, target):
        image = torchvision.transforms.functional.to_tensor(image)
        return image, target

class RandomHorizontalFlip:
    def __init__(self, prob=0.5):
        self.prob = prob

    def __call__(self, image, target):
        if random.random() < self.prob:
            image = torchvision.transforms.functional.hflip(image)

            # Need to flip the boxes too
            boxes = target["boxes"]
            width = image.shape[-1]
            boxes[:, [0, 2]] = width - boxes[:, [2, 0]]  # Flip x-coordinates
            target["boxes"] = boxes

        return image, target

# Define transformations for object detection
def get_transform(train):
    transforms_list = []
    # Convert PIL image to tensor
    transforms_list.append(ToTensor())
    if train:
        # Add data augmentation for training
        transforms_list.append(RandomHorizontalFlip(0.5))

    return Compose(transforms_list)

# Custom dataset class for VOC format
class TrafficSignDataset(torch.utils.data.Dataset):
    def __init__(self, root_dir, transforms=None):
        self.root_dir = root_dir
        self.transforms = transforms

        # Get all jpg files
        self.imgs = [f for f in os.listdir(root_dir) if f.endswith('.jpg')]
        self.imgs.sort()  # Ensure consistent ordering

        # Map class names to indices
        self.class_to_idx = {cls_name: i+1 for i, cls_name in enumerate(CLASS_NAMES)}

    def __getitem__(self, idx):
        # Load image
        img_path = os.path.join(self.root_dir, self.imgs[idx])
        img = Image.open(img_path).convert("RGB")

        # Get corresponding XML file
        xml_path = os.path.join(self.root_dir, self.imgs[idx].replace('.jpg', '.xml'))

        # Parse XML
        tree = ET.parse(xml_path)
        root = tree.getroot()

        boxes = []
        labels = []
        areas = []

        # Extract annotation info
        for obj in root.findall('object'):
            name = obj.find('name').text
            if name not in self.class_to_idx:
                continue  # Skip if class not in our list

            label = self.class_to_idx[name]

            bbox = obj.find('bndbox')
            xmin = float(bbox.find('xmin').text)
            ymin = float(bbox.find('ymin').text)
            xmax = float(bbox.find('xmax').text)
            ymax = float(bbox.find('ymax').text)

            # Add box only if it's valid
            if xmax > xmin and ymax > ymin:
                boxes.append([xmin, ymin, xmax, ymax])
                labels.append(label)
                areas.append((xmax - xmin) * (ymax - ymin))

        # If no valid boxes found, create a dummy box to avoid errors
        if len(boxes) == 0:
            boxes = torch.zeros((0, 4), dtype=torch.float32)
            labels = torch.zeros((0), dtype=torch.int64)
            areas = torch.zeros((0), dtype=torch.float32)
            iscrowd = torch.zeros((0), dtype=torch.int64)
        else:
            # Convert to tensors
            boxes = torch.as_tensor(boxes, dtype=torch.float32)
            labels = torch.as_tensor(labels, dtype=torch.int64)
            areas = torch.as_tensor(areas, dtype=torch.float32)
            iscrowd = torch.zeros((len(labels),), dtype=torch.int64)

        image_id = torch.tensor([idx])

        # Create target dictionary
        target = {}
        target["boxes"] = boxes
        target["labels"] = labels
        target["image_id"] = image_id
        target["area"] = areas
        target["iscrowd"] = iscrowd

        # Apply transformations if any
        if self.transforms is not None:
            img, target = self.transforms(img, target)

        return img, target

    def __len__(self):
        return len(self.imgs)

# Collate function for data loader
def collate_fn(batch):
    return tuple(zip(*batch))

# Load datasets
print("Loading datasets...")
train_dataset = TrafficSignDataset(train_dir, transforms=get_transform(train=True))
valid_dataset = TrafficSignDataset(valid_dir, transforms=get_transform(train=False))
test_dataset = TrafficSignDataset(test_dir, transforms=get_transform(train=False))

print(f"Train dataset size: {len(train_dataset)}")
print(f"Validation dataset size: {len(valid_dataset)}")
print(f"Test dataset size: {len(test_dataset)}")

# Create data loaders
train_loader = DataLoader(
    train_dataset,
    batch_size=CONFIG['batch_size'],
    shuffle=True,
    num_workers=CONFIG['num_workers'],
    collate_fn=collate_fn
)

valid_loader = DataLoader(
    valid_dataset,
    batch_size=CONFIG['batch_size'],
    shuffle=False,
    num_workers=CONFIG['num_workers'],
    collate_fn=collate_fn
)

test_loader = DataLoader(
    test_dataset,
    batch_size=CONFIG['batch_size'],
    shuffle=False,
    num_workers=CONFIG['num_workers'],
    collate_fn=collate_fn
)

# Load pretrained SSD300 model with updated torchvision API
print("Loading SSD300 model...")
num_classes = len(CLASS_NAMES) + 1  # +1 for background class

# For newer torchvision versions
model = torchvision.models.detection.ssd300_vgg16(
    weights=SSD300_VGG16_Weights.DEFAULT
)

# Replace the classification and regression heads for our number of classes
from torchvision.models.detection.ssd import SSDHead
in_channels = [512, 1024, 512, 256, 256, 256]  # Default for SSD300
num_anchors = [4, 6, 6, 6, 4, 4]  # Default for SSD300

# Create new head with our number of classes
model.head = SSDHead(in_channels, num_anchors, num_classes)

# Move model to device
model.to(device)

# Set up optimizer
params = [p for p in model.parameters() if p.requires_grad]
optimizer = optim.SGD(params, lr=CONFIG['learning_rate'], momentum=0.9, weight_decay=0.0005)
lr_scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)

# Function to calculate mAP using PyTorch's evaluation tools
def calculate_map(model, data_loader, device):
    model.eval()

    # Prepare metrics accumulation
    all_predictions = []
    all_targets = []

    with torch.no_grad():
        for images, targets in tqdm(data_loader, desc="Evaluating"):
            images = list(img.to(device) for img in images)
            targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

            outputs = model(images)

            # Store predictions and targets for calculating metrics
            all_predictions.extend(outputs)
            all_targets.extend(targets)

    # Calculate mAP and F1 using COCO metrics
    from pycocotools.cocoeval import COCOeval
    from pycocotools.coco import COCO
    import json
    import tempfile

    # Create temporary directory for COCO JSON files
    temp_dir = tempfile.mkdtemp()

    # Convert predictions and targets to COCO format
    coco_gt = convert_to_coco_format(all_targets)
    coco_dt = convert_to_coco_format(all_predictions, is_gt=False)

    # Save temporary JSON files
    gt_file = os.path.join(temp_dir, 'coco_gt.json')
    dt_file = os.path.join(temp_dir, 'coco_dt.json')

    with open(gt_file, 'w') as f:
        json.dump(coco_gt, f)

    with open(dt_file, 'w') as f:
        json.dump(coco_dt, f)

    # Load as COCO objects
    coco_gt = COCO(gt_file)

    # Check if we have any predictions
    if len(coco_dt) > 0:
        coco_dt = coco_gt.loadRes(dt_file)

        # Create COCO evaluator
        coco_eval = COCOeval(coco_gt, coco_dt, 'bbox')
        coco_eval.evaluate()
        coco_eval.accumulate()
        coco_eval.summarize()

        # Extract mAP
        map_50_95 = coco_eval.stats[0]  # AP@[.5:.95]
        map_50 = coco_eval.stats[1]     # AP@.50
    else:
        print("No predictions to evaluate!")
        map_50_95 = 0.0
        map_50 = 0.0

    # Calculate F1 score (custom implementation)
    f1_score = calculate_f1_score(all_predictions, all_targets)

    return map_50, f1_score

# Helper function to convert to COCO format
def convert_to_coco_format(data_list, is_gt=True):
    if is_gt:
        coco_dict = {
            "images": [],
            "annotations": [],
            "categories": [{"id": i+1, "name": name} for i, name in enumerate(CLASS_NAMES)]
        }

        annotation_id = 1

        for img_id, data in enumerate(data_list):
            # Add image info
            coco_dict["images"].append({
                "id": img_id,
                "width": 300,  # Assuming SSD300 input size
                "height": 300
            })

            # Ground truth
            boxes = data["boxes"].cpu().numpy()
            labels = data["labels"].cpu().numpy()

            for box_id, (box, label) in enumerate(zip(boxes, labels)):
                x1, y1, x2, y2 = box
                width = x2 - x1
                height = y2 - y1

                coco_dict["annotations"].append({
                    "id": annotation_id,
                    "image_id": img_id,
                    "category_id": int(label),
                    "bbox": [float(x1), float(y1), float(width), float(height)],
                    "area": float(width * height),
                    "iscrowd": 0
                })
                annotation_id += 1

        return coco_dict
    else:
        # Predictions
        coco_predictions = []

        for img_id, data in enumerate(data_list):
            if "boxes" in data and len(data["boxes"]) > 0:
                boxes = data["boxes"].cpu().numpy()
                scores = data["scores"].cpu().numpy()
                labels = data["labels"].cpu().numpy()

                for box_id, (box, score, label) in enumerate(zip(boxes, scores, labels)):
                    x1, y1, x2, y2 = box
                    width = x2 - x1
                    height = y2 - y1

                    coco_predictions.append({
                        "image_id": img_id,
                        "category_id": int(label),
                        "bbox": [float(x1), float(y1), float(width), float(height)],
                        "score": float(score)
                    })

        return coco_predictions

# Function to calculate F1 score
def calculate_f1_score(predictions, targets, iou_threshold=0.5, score_threshold=0.5):
    total_tp = 0
    total_fp = 0
    total_fn = 0

    # Process each image
    for pred, target in zip(predictions, targets):
        pred_boxes = pred["boxes"].cpu().numpy()
        pred_scores = pred["scores"].cpu().numpy()
        pred_labels = pred["labels"].cpu().numpy()

        # Filter predictions by confidence score
        keep = pred_scores > score_threshold
        pred_boxes = pred_boxes[keep]
        pred_labels = pred_labels[keep]

        gt_boxes = target["boxes"].cpu().numpy()
        gt_labels = target["labels"].cpu().numpy()

        # Track which ground truth boxes have been matched
        gt_matched = [False] * len(gt_boxes)

        # For each prediction, find if it matches a ground truth box
        for pred_box, pred_label in zip(pred_boxes, pred_labels):
            matched = False
            best_iou = 0
            best_gt_idx = -1

            # Compare with each ground truth box
            for gt_idx, (gt_box, gt_label) in enumerate(zip(gt_boxes, gt_labels)):
                # Skip already matched ground truths or different classes
                if gt_matched[gt_idx] or pred_label != gt_label:
                    continue

                # Calculate IoU
                iou = calculate_iou(pred_box, gt_box)
                if iou > iou_threshold and iou > best_iou:
                    best_iou = iou
                    best_gt_idx = gt_idx
                    matched = True

            if matched:
                total_tp += 1
                gt_matched[best_gt_idx] = True
            else:
                total_fp += 1

        # Count false negatives (ground truth boxes that weren't matched)
        total_fn += sum(1 for m in gt_matched if not m)

    # Calculate precision and recall
    precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) > 0 else 0
    recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) > 0 else 0

    # Calculate F1 score
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    return f1

# Calculate IoU between two boxes
def calculate_iou(box1, box2):
    # box format: [x1, y1, x2, y2]
    x1_1, y1_1, x2_1, y2_1 = box1
    x1_2, y1_2, x2_2, y2_2 = box2

    # Intersection coordinates
    x1_i = max(x1_1, x1_2)
    y1_i = max(y1_1, y1_2)
    x2_i = min(x2_1, x2_2)
    y2_i = min(y2_1, y2_2)

    # Check if there is an intersection
    if x2_i < x1_i or y2_i < y1_i:
        return 0.0

    # Calculate intersection area
    intersection = (x2_i - x1_i) * (y2_i - y1_i)

    # Calculate union area
    area1 = (x2_1 - x1_1) * (y2_1 - y1_1)
    area2 = (x2_2 - x1_2) * (y2_2 - y1_2)
    union = area1 + area2 - intersection

    # Calculate IoU
    iou = intersection / union if union > 0 else 0
    return iou

# Training function
def train_one_epoch(model, optimizer, data_loader, device):
    model.train()
    total_loss = 0

    for images, targets in tqdm(data_loader, desc="Training"):
        images = list(img.to(device) for img in images)
        targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

        # Forward pass
        loss_dict = model(images, targets)
        losses = sum(loss for loss in loss_dict.values())

        # Backward pass
        optimizer.zero_grad()
        losses.backward()
        optimizer.step()

        total_loss += losses.item()

    return total_loss / len(data_loader)

# Main training and evaluation loop
def train_and_evaluate(model, optimizer, train_loader, valid_loader, test_loader,
                      config, device):
    # Initialize metrics tracking
    best_map = 0.0
    best_f1 = 0.0
    best_epoch = 0
    epochs_without_improvement = 0

    train_losses = []
    valid_maps = []
    valid_f1s = []

    lr_scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)

    print("Starting training...")
    for epoch in range(config['epochs']):
        print(f"\nEpoch {epoch+1}/{config['epochs']}")

        # Train one epoch
        train_loss = train_one_epoch(model, optimizer, train_loader, device)
        train_losses.append(train_loss)

        # Evaluate on validation set
        print("Evaluating on validation set...")
        valid_map, valid_f1 = calculate_map(model, valid_loader, device)
        valid_maps.append(valid_map)
        valid_f1s.append(valid_f1)

        print(f"Epoch {epoch+1} - Train Loss: {train_loss:.4f}, Valid mAP@0.5: {valid_map:.4f}, Valid F1: {valid_f1:.4f}")

        # Update learning rate
        lr_scheduler.step()

        # Check if this is the best model
        improved = False
        if valid_map > best_map or valid_f1 > best_f1:
            if valid_map > best_map:
                best_map = valid_map
            if valid_f1 > best_f1:
                best_f1 = valid_f1

            best_epoch = epoch
            improved = True
            epochs_without_improvement = 0

            # Save best model
            print(f"Saving best model at epoch {epoch+1}...")
            torch.save({
    'epoch': epoch,
    'model_state_dict': model.state_dict(),
    'optimizer_state_dict': optimizer.state_dict(),
    'map': float(valid_map),  # Convert to Python float
    'f1': float(valid_f1),    # Convert to Python float
}, os.path.join(config['results_dir'], 'best.pth'), _use_new_zipfile_serialization=True)
        else:
            epochs_without_improvement += 1
            print(f"No improvement for {epochs_without_improvement} epochs.")

        # Early stopping
        if epochs_without_improvement >= config['early_stopping_patience']:
            print(f"Early stopping triggered after {epoch+1} epochs.")
            break

    print(f"Training complete. Best model was at epoch {best_epoch+1} with mAP: {best_map:.4f} and F1: {best_f1:.4f}")

    # Plot training curves
    plot_training_curves(train_losses, valid_maps, valid_f1s, config)

    # Evaluate best model on test set
    print("Loading best model for test evaluation...")
    checkpoint = torch.load(os.path.join(config['results_dir'], 'best.pth'), weights_only=False)
    model.load_state_dict(checkpoint['model_state_dict'])

    print("Evaluating on test set...")
    test_map, test_f1 = calculate_map(model, test_loader, device)
    print(f"Test mAP@0.5: {test_map:.4f}, Test F1: {test_f1:.4f}")

    # Save final metrics
    metrics_df = pd.DataFrame({
        'Metric': ['Best Validation mAP@0.5', 'Best Validation F1',
                   'Test mAP@0.5', 'Test F1',
                   'Best Epoch', 'Total Epochs'],
        'Value': [best_map, best_f1, test_map, test_f1, best_epoch+1, epoch+1]
    })
    metrics_df.to_csv(os.path.join(config['results_dir'], 'metrics.csv'), index=False)

    return best_map, best_f1, test_map, test_f1

# Function to plot and save training curves
def plot_training_curves(train_losses, valid_maps, valid_f1s, config):
    # Plot training loss
    plt.figure(figsize=(12, 4))

    # Plot 1: Training Loss
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, 'b-', label='Training Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training Loss per Epoch')
    plt.grid(True)
    plt.legend()

    # Plot 2: Validation metrics
    plt.subplot(1, 2, 2)
    plt.plot(valid_maps, 'r-', label='mAP@0.5')
    plt.plot(valid_f1s, 'g-', label='F1 Score')
    plt.xlabel('Epoch')
    plt.ylabel('Score')
    plt.title('Validation Metrics per Epoch')
    plt.grid(True)
    plt.legend()

    plt.tight_layout()
    plt.savefig(os.path.join(config['results_dir'], 'training_curves.png'))

    # Save individual plots for mAP and F1
    plt.figure(figsize=(10, 6))
    plt.plot(valid_maps, 'r-o')
    plt.xlabel('Epoch')
    plt.ylabel('mAP@0.5')
    plt.title('mAP@0.5 per Epoch')
    plt.grid(True)
    plt.savefig(os.path.join(config['results_dir'], 'mAP_plot.png'))

    plt.figure(figsize=(10, 6))
    plt.plot(valid_f1s, 'g-o')
    plt.xlabel('Epoch')
    plt.ylabel('F1 Score')
    plt.title('F1 Score per Epoch')
    plt.grid(True)
    plt.savefig(os.path.join(config['results_dir'], 'F1_plot.png'))

# Function to visualize some predictions
def visualize_predictions(model, data_loader, device, num_images=5):
    model.eval()
    images_so_far = 0
    fig = plt.figure(figsize=(15, 12))

    with torch.no_grad():
        for images, targets in data_loader:
            images = list(img.to(device) for img in images)

            # Get predictions
            outputs = model(images)

            for i, (img, output, target) in enumerate(zip(images, outputs, targets)):
                if images_so_far >= num_images:
                    plt.tight_layout()
                    plt.savefig(os.path.join(CONFIG['results_dir'], 'predictions.png'))
                    plt.show()
                    return

                images_so_far += 1
                ax = fig.add_subplot(num_images, 1, images_so_far)

                # Convert tensor to numpy for visualization
                img_np = img.cpu().permute(1, 2, 0).numpy()
                img_np = np.clip(img_np, 0, 1)  # Clip values to [0, 1] range

                ax.imshow(img_np)

                # Draw ground truth boxes in green
                gt_boxes = target['boxes'].cpu().numpy()
                gt_labels = target['labels'].cpu().numpy()

                for box, label_idx in zip(gt_boxes, gt_labels):
                    rect = plt.Rectangle((box[0], box[1]), box[2]-box[0], box[3]-box[1],
                                        fill=False, color='green', linewidth=2)
                    ax.add_patch(rect)
                    class_name = CLASS_NAMES[label_idx-1] if label_idx-1 < len(CLASS_NAMES) else f"Unknown {label_idx}"
                    ax.text(box[0], box[1], f"GT: {class_name}",
                           bbox=dict(facecolor='green', alpha=0.5), fontsize=8)

                # Draw predicted boxes in red (only those with score > 0.5)
                if len(output['boxes']) > 0:
                    pred_boxes = output['boxes'].cpu().numpy()
                    pred_labels = output['labels'].cpu().numpy()
                    pred_scores = output['scores'].cpu().numpy()

                    for box, label_idx, score in zip(pred_boxes, pred_labels, pred_scores):
                        if score > 0.5:  # Only show predictions with score > 0.5
                            rect = plt.Rectangle((box[0], box[1]), box[2]-box[0], box[3]-box[1],
                                                fill=False, color='red', linewidth=2)
                            ax.add_patch(rect)
                            class_name = CLASS_NAMES[label_idx-1] if label_idx-1 < len(CLASS_NAMES) else f"Unknown {label_idx}"
                            ax.text(box[0], box[3] + 10,
                                   f"Pred: {class_name} ({score:.2f})",
                                   bbox=dict(facecolor='red', alpha=0.5), fontsize=8)

                ax.set_title(f"Image {images_so_far}")
                ax.axis('off')

    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG['results_dir'], 'predictions.png'))
    plt.show()

# Run the training and evaluation
print("Starting the training and evaluation process...")
best_val_map, best_val_f1, test_map, test_f1 = train_and_evaluate(
    model, optimizer, train_loader, valid_loader, test_loader, CONFIG, device
)

print("\nTraining and Evaluation Complete!")
print(f"Best validation mAP@0.5: {best_val_map:.4f}")
print(f"Best validation F1 score: {best_val_f1:.4f}")
print(f"Test mAP@0.5: {test_map:.4f}")
print(f"Test F1 score: {test_f1:.4f}")
print(f"Results saved to {CONFIG['results_dir']}")

# Visualize some test predictions using the best model
print("Visualizing some test predictions...")
visualize_predictions(model, test_loader, device, num_images=5)
```
